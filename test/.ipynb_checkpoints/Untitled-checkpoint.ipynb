```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import models, datasets, transforms
import matplotlib.pyplot as plt
import pickle
import urllib.request
from elephantreid.identification.network import *
import numpy as np
from PIL import Image
from pathlib import Path
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
import seaborn as sns

#PARAMETERS
seed = 42
out_dir = "resultats2"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
num_classes = 12
#CHARGEMENT DU MODELE
base_model = models.resnet50(weights='DEFAULT')
#ajouter un block de 3 couches linéaires : tête de classification à entraîner
classifier_name = "fc"
dim_in = 2048
setattr(base_model,classifier_name,ClassificationHead(dim_out=num_classes,dim_in=dim_in))
assert(getattr(base_model,classifier_name).out_features==num_classes)
```

```python
base_model = models.resnet50(weights='DEFAULT')

#ajouter un block de 3 couches linéaires : tête de classification à entraîner
classifier_name = "fc"
dim_in = 2048
setattr(base_model,classifier_name,ClassificationHead(dim_out=num_classes,dim_in=dim_in))
assert(getattr(base_model,classifier_name).out_features==num_classes)
```

```python
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
g = torch.Generator().manual_seed(seed)
num_epoch = 10
num_train = len(train_features_dataset)
num_val = len(val_features_dataset)
batch_size = 42
criterion = nn.CrossEntropyLoss()
for param in base_model.parameters():
    param.requires_grad = False
for param in base_model.fc.parameters():
    param.requires_grad = True
optimizer = torch.optim.Adam(base_model.fc.parameters(),lr=0.0005)
train_error = [] ; val_error =[]
train_accuracy = [] ; val_accuracy = []
train_loader = DataLoader(train_features_dataset,shuffle=True,generator=g,batch_size=batch_size)
val_loader = DataLoader(val_features_dataset,batch_size=batch_size)
all_predicted, all_labels = [],[]
for epoch in range(num_epoch):
    ### TRAINING
    base_model.eval() # freeze the other parameters
    base_model.fc.train() # train last layer
    error = 0 ; corrects =0
    for feat,label in train_loader:
        output = base_model.fc(feat)
        loss = criterion(output,label)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        error+= loss.item()
        _,predicted = torch.max(output,1)
        corrects += (predicted==label).sum()
    error /= len(train_loader)
    train_error.append(error)
    train_accuracy.append(corrects*100/num_train)
    ### VALIDATION 
    base_model.eval()
    error = 0 ; corrects =0
    with torch.no_grad():
        for feat,label in val_loader:
            output = base_model.fc(feat)
            error+= criterion(output,label).item()
            _,predicted = torch.max(output,1)
            corrects += (predicted==label).sum()
            if epoch==num_epoch-1:
                all_predicted.append(predicted.cpu());all_labels.append(label.cpu())
        error /= len(val_loader)
        val_error.append(error)
        val_accuracy.append(corrects*100/num_val)
plt.figure(figsize=(10,10))
plt.subplot(2,1,1)
plt.plot(range(1,num_epoch+1),train_error,color="blue",label="train")
plt.plot(range(1,num_epoch+1),val_error,color="orange",label="validation")
plt.title("Loss")
plt.legend()
plt.subplot(2,1,2)
plt.plot(range(1,num_epoch+1),train_accuracy,color="blue",label="train")
plt.plot(range(1,num_epoch+1),val_accuracy,color="orange",label="validation")
plt.title("Accuracy")
```

```python
def save_grad_cam1(model: models.ResNet, classes : dict, image: torch.tensor, out_dir:str) -> None:
    """
    Displays Grad-CAM for top-3 classes + Lowest class.
    """
    activations, gradients = None, None

    ### dernière couche de convolution
    last_name = None
    last_conv = None
    flag = False
    for name, m in model.named_modules():
        if isinstance(m,nn.Conv2d):
            last_name = name
            last_conv = m
    for name, param in model.named_parameters():
        name = '.'.join(name.split('.')[:-1])
        if name == last_name:
            flag=True
        param.requires_grad = flag
    def forward_hook(module,input,output):
        ### hook qui affecte à l'activation de la couche
        nonlocal activations
        activations = output
    
    def backward_hook(module,grad_input,grad_output):
        ### hook qui affecte à gradients la dérivée de la sortie par rapport à l'activation de la couche module
        nonlocal gradients
        gradients = grad_output[0] #tuple gradient par rapport aux sorties (ici une)
    
    handle_forward = last_conv.register_forward_hook(forward_hook)
    handle_backward = last_conv.register_full_backward_hook(backward_hook)
    model.eval()
    output = model(image)
    values,indexes = torch.topk(output, 3)
    v, i = torch.topk(output,1,largest=False)
    indexes = indexes.tolist()[0] +  i.tolist()[0]
    values = values.tolist()[0] +  v.tolist()[0]
    plot_image = (image[0].permute(1,2,0).detach().cpu()).clone()
    plot_image-= plot_image.min()
    plot_image/=plot_image.max()
    extent = 0,224,0,224
    sp = 1
    plt.figure(figsize=(4,4))
    for i,v in zip(indexes,values):
        model.zero_grad()
        output[0,i].backward(retain_graph=True)
        alpha = gradients.mean(dim=(2,3),keepdim=True)
        grad_cam = F.relu((alpha*activations).sum(dim=(1),keepdim=True))
        grad_cam -= grad_cam.min()
        grad_cam /= grad_cam.max()
        grad_cam = grad_cam[0,0].detach().cpu()
        plt.subplot(2,2,sp)
        plt.imshow(plot_image,extent=extent)
        plt.imshow(grad_cam,cmap="viridis",alpha=0.5,interpolation="bilinear",extent=extent)
        plt.title(f"{classes[i]} : {v:.2f}")
        sp+=1
    #save figure
    if not os.listdir(out_dir):
        crop = "0"
    else:
        crop = str(int(sorted(os.listdir(out_dir))[-1].split('.')[0])+1)
    out_path = os.path.join(out_dir,crop)
    plt.tight_layout()
    plt.savefig(out_path, dpi=300, bbox_inches="tight")
    plt.close()
    handle_forward.remove(), handle_backward.remove()
```

```python
from pytorch_metric_learning.losses import ArcFaceLoss
embedding_dim = 512
arcface = ArcFaceLoss(num_classes=num_classes,embedding_size=embedding_dim)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
train_feat = precompute_features(model,train_data,device=device)
val_feat = precompute_features(model,val_data,device=device)
```

```python
X_train, y_train = train_feat.tensors
X_val, y_val = val_feat.tensors
```

```python
num_classes = 12
embedding_dim = 512
epochs = 10

class ProjectionModel(nn.Modules):
    def __init__(self,in_dim,projection_dim=512):
        self.projection = nn.Sequential([nn.Linear(in_dim,projection_dim),nn.BatchNorm1d(projection_dim)])
    def forward(self,x):
        emb = self.projection(x)
        emb = F.normalize(emb,p=2,dim=1)
                              
model = ProjectionModel(
    projection_dim=embedding_dim
).to(device)

arcface = ArcFaceLoss(num_classes=num_classes,embedding_size=embedding_dim)

optimizer = torch.optim.AdamW(
    list(model.parameters()) +
    list(arcface.parameters()),
    lr=1e-4,
    weight_decay=1e-4
)

# =====================================================
# Training
# =====================================================

for epoch in range(epochs):

    model.train()

    running_loss = 0.0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        embeddings = model(images)

        logits = arcface(
            embeddings,
            labels
        )

        loss = F.cross_entropy(
            logits,
            labels
        )

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

    print(
        f"Epoch {epoch+1}/{epochs} "
        f"Loss: {running_loss/len(train_loader):.4f}"
    )

# =====================================================
# Example retrieval
# =====================================================

model.eval()

with torch.no_grad():

    embeddings = []

    labels_all = []

    for images, labels in train_loader:

        images = images.to(device)

        emb = model(images)

        embeddings.append(emb.cpu())

        labels_all.append(labels)

    embeddings = torch.cat(embeddings)
    labels_all = torch.cat(labels_all)

# cosine similarity matrix
similarity = embeddings @ embeddings.T

print("Embedding shape:", embeddings.shape)
print("Similarity matrix shape:", similarity.shape)
```
