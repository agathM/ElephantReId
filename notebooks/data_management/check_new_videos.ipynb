```python
import pandas as pd
import os
file_path = "../../database_info/Beach_elephants_10_06.csv"
dt = pd.read_csv(file_path)
dt = dt.filter(['Name','indiv_name'])
videos_path = "../data/beach_individuals"
new_videos = set()
for indiv in os.listdir(videos_path):
    indiv_path = os.path.join(videos_path,indiv)
    videos = set(os.listdir(indiv_path))
    videos_dt = set(dt[dt['indiv_name']==indiv]['Name'])
    new_videos.update(videos_dt - videos)
print(new_videos)
```
