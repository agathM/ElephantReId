```python
import pandas as pd
import os, shutil
file_path = "../../data/input/DB_SEB_VD.ods"
dt = pd.read_excel(file_path,engine='odf')
dt = dt.filter(['clip','mode']).set_index('clip').to_dict()['mode']

videos_path = "../../data/input/cave_individuals"
for indiv in os.listdir(videos_path):
    indiv_path = os.path.join(videos_path,indiv)
    os.makedirs(os.path.join(indiv_path,'Night'),exist_ok=True)
    os.makedirs(os.path.join(indiv_path,'Day'),exist_ok=True)
    for video in os.listdir(indiv_path):
        if video in dt.keys():
            mode = dt[video]
            src = os.path.join(indiv_path,video)
            dst = os.path.join(indiv_path,mode,video)
            shutil.move(src=src,dst=dst)
```

write summary per individuals

```python
import pandas as pd
import os

videos_path = '../../data/input/beach_individuals'
rows = []

for indiv in os.listdir(videos_path):
    indiv_path = os.path.join(videos_path, indiv)
    for mode in os.listdir(indiv_path):
        mode_path = os.path.join(indiv_path, mode)
        videos = os.listdir(mode_path)
        rows.append({
            "individual": indiv,
            "mode": mode,
            "number_videos": len(videos)
        })
df1 = pd.DataFrame(rows)

videos_path = "../../data/input/cave_individuals"
for indiv in os.listdir(videos_path):
    indiv_path = os.path.join(videos_path, indiv)
    for mode in os.listdir(indiv_path):
        mode_path = os.path.join(indiv_path, mode)
        videos = os.listdir(mode_path)
        rows.append({
            "individual": indiv,
            "mode": mode,
            "number_videos": len(videos)
        })
df2 = pd.DataFrame(rows)
with pd.ExcelWriter("../../data/input/database.ods",engine='odf') as writer:
    df1.to_excel(writer,sheet_name="beach_individuals",index=False)
    df2.to_excel(writer,sheet_name="cave_individuals",index=False)
```
